```python
from currency_symbols import CurrencySymbols
class BankAccountSystem:
  def __init__(self,name,amount):
    self.__name=name
    self.__amount=amount

  def deposite(self,balance):
    self.__amount+=balance
    print(f"You deposite {balance}{CurrencySymbols.get_symbol('INR')}.")
    print("Successfully deposite...")

  def withdraw(self,amount):
    if amount<0 or amount>self.__amount:
      print("Invalid amount...")
    else:
      self.__amount-=amount
      print("Successfully withdraw...")

  def checkbalance(self):
    print(f"You have {self.__amount} in your Account.")

aniket=BankAccountSystem('Aniket',1000000000)
aniket.deposite(200000000)
aniket.withdraw(1)
aniket.checkbalance()


```

```python
class StudentClassManager:

  def __init__(self,name,roll,marks):
    self.__name=name
    self.__roll=roll
    self.__marks=marks

  def calc_percentage(self):
    total=sum(self.__marks)
    return total/len(self.__marks)

  def studentdetails(self):
    print(f"Name:{self.__name}\nRoll No:{self.__roll}\nPercentage:{self.calc_percentage()}%")

aniket=StudentClassManager("Aniket",20233052,[99,99,85,89,71])
aniket.studentdetails()
```

```python
class LibraryBooks:
  def __init__(self):
    self.book_list=[]

  def AddBook(self,book):
    self.book_list.append(book)
    print(f"{book} added in Library.")

  def IssueBook(self,book):
    if book in self.book_list:
      self.book_list.remove(book)
      print(f"You issued {book}")
    else:
      print(f"Sorry!Book is not available.")

  def ReturnBook(self,book):
    self.book_list.append(book)
    print(f"Thank You!You return {book}")

  def CheckAvailability(self,book_name):
    if book_name in self.book_list:
      print("Book is Available in Library.")
    else:
      print("Sorry,Book is not available in Library.")

love_is_real=LibraryBooks()
love_is_real.AddBook("Rich Dad ,Poor Dad")
love_is_real.AddBook("Business School")
love_is_real.ReturnBook("Business School")
love_is_real.CheckAvailability("Business School")
```

```python
import json

open_file_name='Data.json'
# save_file_name='Save_data.json'

def load_data():
  try:
    with open(open_file_name,'r') as f:
      return json.load(f)
  except FileNotFoundError as e:
    return {}

def save_data(students):
    with open(open_file_name,'w') as f:
      json.dump(students,f)

def add_new_student(students):
  try:
    name=input("Enter Student Name:")
    roll=int(input("Enter roll number of Student:"))
    marks=list(map(int,input("Enter marks with one space:").split()))

    students[roll]={
        "Name":name,
        "Marks":marks
    }
    save_data(students)
    print("Save data Successfully.")
  except ValueError:
    print("Data not saved.")

def search_student(students):
  roll=int(input("Enter the Student Roll Number:"))
  if roll in students:
    print(students[roll])
  else:
    print("Student's Not found..")

def update_marks(students):
  roll=int(input("Enter the Roll Number of Student:"))
  if roll in students:
    try:
      marks=list(map(int,input("Enter the marks with space").split()))
      students[roll]['Marks']=marks
      save_data(students)
      print("Marks updated.")
    except:
      print("Value error")
  else:
    print("Roll no not found.")

def delete_student(students):
  roll=int(input("Enter Roll No:"))
  if roll in students:
    del students[roll]
    save_data(students)
  else:
    print("Roll no exist.")

def calculate_grade(avg):
  if avg>=90:
    return "A"
  elif avg>=75:
    return 'B'
  elif avg>=60:
    return 'C'
  else:
    return 'D'

def view_student_details(students):
  for roll,data in students.items():
    print(data)
    marks=data["Marks"]
    total=sum(marks)
    avarage=total/len(marks)
    grade=calculate_grade(avarage)

    print(f"\n Roll:{roll}")
    print(f"Name :{data['Name']}")
    print(f"Marks:{marks}")
    print(f"Total:{total}")
    print(f"Avarage:{avarage}")
    print(f"Grade:{grade}")

def main():
  students=load_data()

  while True:
    print('1 for Add New Student:')
    print('2 for Search Student:')
    print('3 for Update Student Marks:')
    print('4 for Delete Student:')
    print('5 for View Student Details:')
    print('6 for Exit:')

    option=int(input("Enter your Option:"))

    if option==1:
      add_new_student(students)
    elif option==2:
      search_student(students)
    elif option==3:
      update_marks(students)
    elif option==4:
      delete_student(students)
    elif option==5:
      view_student_details(students)
    else:
      break

main()
```

```python
class Atm_Machine:

  def __init__(self,name,pin,balance=0):
    self.__name=name
    self.__pin=pin
    self.__balance=balance

  def pin_varification(self,pin):

    if self.__pin == pin:
      return True

    return False

  def deposite(self,pin,amount):

        if self.pin_varification(pin):
          self.__balance+=amount
          print("Amount Deposite.")

        else:
          print("Invalid Pin.")

  def withdraw(self,pin,amount):

    if self.pin_varification(pin):
      if self.__balance<0 or self.__balance<amount:
        print("Insufficient Balance.")
      else:
        self.__balance-=amount
    else:
      print('Invalid Pin.')

  def Balance_inquiry(self,pin):

    if self.pin_varification(pin):
      print("Name:",self.__name)
      print("Balance",self.__balance)
    else:
      print('Invalid Pin')

fa=Atm_Machine('Aniket',20233052,1000000)
fa.deposite(20233052,10000)
fa.Balance_inquiry(20233052)
```

```python
class Employee:
  def __init__(self,name,salary,emp_id):
    self.__name=name
    self.__salary=salary
    self.__emp_id=emp_id

  def get_name(self):
    return self.__name

  def get_salary(self):
    return self.__salary

  def __employee_info(self):
    print(f"Name of Employee: {self.__name}")
    print(f"Salary of Employee: {self.__salary}")
    print(f"Emp. Id :{self.__emp_id}")

  def get_info(self):
    return self.__employee_info()

class Manager(Employee):
  def __init__(self, name, salary,emp_id,bonus):
    super().__init__(name, salary,emp_id)
    self.__bonus=bonus

  def calculate_salary(self):
    self.get_info()
    return f"{self.get_name()}'s Final Salary:{self.get_salary()+self.__bonus} "

class Developer(Employee):
  def __init__(self, name, salary,emp_id,bonus):
    super().__init__(name, salary,emp_id)
    self.__bonus=bonus

  def calculate_salary(self):
    self.get_info()
    return  f"{self.get_name()}'s Final Salary:{self.get_salary()+self.__bonus} "


aniket=Manager('Aniket',200000,20233052,200000)
aniket.get_info()
aman=Developer('Aman',10000,20233042,1000)
print(aman.calculate_salary())
```
